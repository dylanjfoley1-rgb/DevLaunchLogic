# DevLaunchLogic
My app helps developers decide to launch a free-to-play or pay-to-play game by analyzing player retention and transaction conversion rates across games of similar genres, pricing, or through their own playtests/demos

```python
from google.colab import drive
drive.mount("/content/drive")

import glob
import os
import time
import pandas as pd

# Look for steam_games.csv in My Drive and in folders up to 3 levels down
found = []
for pattern in ["", "*/", "*/*/", "*/*/*/"]:
    found = found + glob.glob("/content/drive/MyDrive/" + pattern + "steam_games.csv")

if len(found) == 0:
    print("❌ Can't find steam_games.csv in your Drive. Upload it next to this notebook and run this cell again.")
else:
    FOLDER = os.path.dirname(found[0]) + "/"
    games = pd.read_csv(FOLDER + "steam_games.csv")
    print(f"✅ Loaded {len(games)} games from:", FOLDER)
```

```python
# Insert target genre
target_genre='RPG'
genre_df=games[games['genres'].str.contains(target_genre, na=False)].copy()
free_df=genre_df[genre_df['price']==0].groupby('release_year')['price'].count()
paid_df=genre_df[genre_df['price']>0].groupby('release_year')['price'].count()
avg_price_by_year=genre_df.groupby('release_year')['price'].mean()
game_count_by_year=genre_df.groupby('release_year')['price'].count()
summary_df=pd.DataFrame({'Average Paid Price':avg_price_by_year, 'Free Games': free_df, 'Paid Games': paid_df, 'Total Games':game_count_by_year}).round(2)
print(summary_df.tail(10))
```
